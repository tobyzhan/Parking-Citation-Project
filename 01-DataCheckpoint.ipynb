## Research Question

**Do poor weather conditions (e.g., excessive rain, low temperature, or high wind speed) significantly increase parking occupancy rates at UCSD parking structures across campus neighborhoods and academic quarters? More specifically, can quarterly parking occupancy rates be predicted from seasonal weather variables such as temperature, precipitation, humidity, and wind speed?**

To address this question, we will analyze parking availability across neighborhoods (e.g., Warren, ERC, Revelle, Muir, Marshall) during Winter, Spring, and Fall 2024, and evaluate how these patterns change under different weather conditions. The primary outcome variable is parking occupancy rate, and explanatory variables include weather metrics, neighborhood, academic quarter, and structure characteristics (e.g., covered vs. surface), as well as parking capacity to normalize availability.

This study is primarily a statistical inference and prediction task, using correlation analysis and regression- or ML-based modeling approaches to quantify weather-related effects and forecast occupancy patterns.

## Background and Prior Work

San Diego is known for its mild, coastal climate, especially in La Jolla, where UC San Diego is located. Nevertheless, storms and bad weather still pose major challenges for city infrastructure. When heavy rain hits the city of San Diego, flooding quickly escalates, affecting traffic, transportation, and parking. San Diego Coastkeeper (2024) explains just how the city's outdated stormwater system struggles to handle intense rainfall, especially within roads and parking lots.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)

Despite San Diego's usually mild climate, these unexpected adverse weather occurrences are growing in importance as climate change continuously affects everywhere around the globe. According to San Diego Coastkeeper and its report, stormwater has become a major public problem due to underinvestment in our drainage system. Parking areas are among the most affected because they collect runoff and are frequently constructed without enough permeable surfaces or drainage capacity. Comprehending this local infrastructure issue clarifies why it is important for us to analyze parking availability at UC San Diego under various weather situations.

Previous studies have demonstrated the significance of weather in forecasting parking demand. In Short-Term Forecasting of Off-Street Parking Occupancy (2021), Elisabeth S. Fokker and Thomas Koch discovered that including weather variables like as temperature, humidity, and precipitation enhanced short-term forecasts of off-street parking occupancy.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) Their research showed that variables other than time or location might account for variations in parking usage. This finding supports the idea that weather data can be a powerful input for parking prediction models.

Predicting parking occupancy with machine learning in the network of things (2020) also reached the same conclusion. When Jesper C. Provoost and Andreas Kamilaris, amongst other researchers, used machine-learning approaches to forecast parking availability, they discovered that models that incorporated weather information performed better.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Their research focused on how people's driving and parking habits may be altered by wind, rain, and temperature changes. These studies offer a solid basis for comprehending the relationship between weather and parking habits.

Unlike previous city-based research conducted, our study focuses on UC San Diego's unique environment, combined with San Diego's changing climate. We will be studying how poor weather, such as rain, high wind, and low temperature, affects parking availability across the campus. This project could help predict when and where parking shortages are most likely to occur. By doing this, we will be able to facilitate more effective parking management on days with poor weather.

## Hypothesis


**We hypothesize that academic quarters with higher frequencies of adverse weather conditions (more days with rain > 5mm, wind speed > 15 mph, or temperature < 10°C or > 30°C) will show significantly higher average parking occupancy rates across UCSD neighborhoods, with the strongest effects in Winter and Fall quarters compared to Spring.**

We predict that covered parking structures will show higher occupancy rates during weather-adverse quarters compared to surface lots, as students and staff seek shelter from rain and extreme temperatures. Additionally, we expect that neighborhoods closer to academic buildings (like Warren and ERC areas) will show greater weather sensitivity than peripheral areas, as people will prioritize minimizing walking distance during poor weather conditions. Our reasoning is based on behavioral economics principles where individuals make cost-benefit decisions about transportation mode and parking location based on weather comfort, with adverse conditions driving increased demand for convenient, covered parking options even at higher costs.
